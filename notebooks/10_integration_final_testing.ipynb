# Phase 10 — Integration & Final Testing

Final verification of the NLP/ML side before Flutter integration.


```python
import os
import joblib
import pandas as pd
import numpy as np

required_paths = [
    "../data/raw/yelp_sentiment_master_dataset.csv",
    "../data/processed/yelp_reviews_cleaned.csv",
    "../data/processed/yelp_reviews_nlp.csv",
    "../model/sentiment_model.joblib",
    "../model/tfidf_vectorizer.joblib"
]

for path in required_paths:
    print("FOUND" if os.path.exists(path) else "MISSING", ":", path)
```

```python
model = joblib.load("../model/sentiment_model.joblib")
vectorizer = joblib.load("../model/tfidf_vectorizer.joblib")

print("Model:", type(model).__name__)
print("Vectorizer:", type(vectorizer).__name__)
print("Classes:", model.classes_)
```

```python
test_reviews = [
    "The food was amazing and the service was excellent.",
    "The food was terrible and I will never come back.",
    "The restaurant was okay and the experience was average."
]

vectors = vectorizer.transform(test_reviews)
predictions = model.predict(vectors)
probabilities = model.predict_proba(vectors)

for review, prediction, probs in zip(test_reviews, predictions, probabilities):
    print("\nReview:", review)
    print("Prediction:", prediction)
    print("Confidence:", round(float(np.max(probs)), 4))
    print("Probabilities:", dict(zip(model.classes_, probs.round(4))))
```
