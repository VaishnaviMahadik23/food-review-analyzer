## Phase 2 — Data Cleaning

### Create a Copy

```python
# Create a copy of the original dataset
df_clean = df.copy()

print("Copy created successfully")
```

### Check Text Values

```python
(df_clean["text"].str.strip() == "").sum()
```

```python
df_clean["text"].isnull().sum()
```

### Convert Date

```python
df_clean["date"] = pd.to_datetime(
    df_clean["date"],
    format="%d-%m-%y %H:%M"
)
```

```python
df_clean["date"].dtype
```

### Check Date Conversion

```python
df_clean["date"].head()
```

### Check Text Data Type

```python
df_clean["text"].dtype
```

```python
df_clean["text"].apply(type).value_counts()
```

### Check Rating Data Types

```python
print(df_clean["stars_review"].dtype)
print(df_clean["stars_business"].dtype)
print(df_clean["review_count"].dtype)
```

### Remove Unnecessary Serial Number

```python
df_clean = df_clean.drop(columns=["Sl No"])
```

### Save Clean Dataset

```python
output_path = "../data/processed/yelp_reviews_cleaned.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully")
```
